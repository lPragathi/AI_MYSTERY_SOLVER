```python
# ============================================================
# 🕵️ AI MYSTERY SOLVER — NO API KEY VERSION
# GOOGLE COLAB — SINGLE CELL
# ============================================================

!pip -q install gradio

import gradio as gr
import re
from datetime import datetime


# ============================================================
# MYSTERY ANALYZER
# ============================================================

def solve_mystery(case_title, clues):

    if not clues.strip():

        return """
# ⚠️ No clues provided

Please enter some clues first.
"""

    text = clues.strip()

    # --------------------------------------------------------
    # Find times
    # --------------------------------------------------------

    time_pattern = r'\b(?:[01]?\d|2[0-3]):[0-5]\d\s*(?:AM|PM)?\b'

    times = re.findall(
        time_pattern,
        text,
        flags=re.IGNORECASE
    )

    # --------------------------------------------------------
    # Split clues into lines
    # --------------------------------------------------------

    lines = [
        line.strip()
        for line in text.split("\n")
        if line.strip()
    ]

    # --------------------------------------------------------
    # TIMELINE
    # --------------------------------------------------------

    timeline = []

    for line in lines:

        match = re.search(
            r'^([^-\n]+)\s*-\s*(.*)$',
            line
        )

        if match:

            time = match.group(1).strip()
            event = match.group(2).strip()

            timeline.append(
                (time, event)
            )

    # --------------------------------------------------------
    # KEYWORDS
    # --------------------------------------------------------

    contradiction_words = [
        "but",
        "however",
        "says",
        "said",
        "different",
        "mismatch",
        "not",
        "no",
        "never",
        "impossible",
        "missing",
        "disappeared",
        "unknown",
        "blurry",
        "didn't",
        "did not",
        "cannot",
        "can't"
    ]

    suspicious_lines = []

    for line in lines:

        lower = line.lower()

        if any(
            word in lower
            for word in contradiction_words
        ):

            suspicious_lines.append(line)

    # --------------------------------------------------------
    # BUILD OUTPUT
    # --------------------------------------------------------

    output = ""

    output += "# 🕵️ AI MYSTERY SOLVER\n\n"

    output += f"## Case: {case_title}\n\n"

    output += (
        "The system analyzed the supplied clues and "
        "identified the following evidence patterns.\n\n"
    )

    output += "---\n\n"

    # --------------------------------------------------------
    # SUMMARY
    # --------------------------------------------------------

    output += "# 🧩 CASE SUMMARY\n\n"

    if len(lines) >= 5:

        output += (
            f"The case contains **{len(lines)} clues** "
            f"and approximately **{len(times)} time references**. "
            "The evidence contains several statements that "
            "should be compared carefully for inconsistencies.\n\n"
        )

    else:

        output += (
            "There are not enough clues for a strong conclusion. "
            "More evidence would improve the analysis.\n\n"
        )

    # --------------------------------------------------------
    # TIMELINE
    # --------------------------------------------------------

    output += "# 🕰️ RECONSTRUCTED TIMELINE\n\n"

    if timeline:

        for time, event in timeline:

            output += (
                f"### ⏰ {time}\n\n"
                f"**{event}**\n\n"
            )

    else:

        output += (
            "No clearly formatted timeline was detected.\n\n"
        )

    output += "---\n\n"

    # --------------------------------------------------------
    # FACTS
    # --------------------------------------------------------

    output += "# 📌 FACTS FROM THE EVIDENCE\n\n"

    for line in lines[:15]:

        output += f"- {line}\n"

    output += "\n---\n\n"

    # --------------------------------------------------------
    # CONTRADICTIONS
    # --------------------------------------------------------

    output += "# ⚠️ POSSIBLE ANOMALIES & CONTRADICTIONS\n\n"

    if suspicious_lines:

        for line in suspicious_lines[:10]:

            output += (
                f"### ⚠️ Evidence requiring attention\n\n"
                f"{line}\n\n"
            )

        output += (
            "These statements are not automatically proof of wrongdoing. "
            "They are simply areas where the evidence may conflict "
            "or require verification.\n\n"
        )

    else:

        output += (
            "No obvious contradiction keywords were detected. "
            "That does not mean the evidence is consistent; "
            "a deeper analysis may still reveal inconsistencies.\n\n"
        )

    output += "---\n\n"

    # --------------------------------------------------------
    # AUTOMATIC PATTERN DETECTION
    # --------------------------------------------------------

    output += "# 🔎 DETECTED PATTERNS\n\n"

    lower_text = text.lower()

    patterns_found = False

    if "missing" in lower_text or "disappeared" in lower_text:

        output += (
            "📦 **Missing item detected:** "
            "The case appears to involve an item that "
            "cannot currently be accounted for.\n\n"
        )

        patterns_found = True

    if "camera" in lower_text or "cctv" in lower_text:

        output += (
            "📹 **Surveillance evidence detected:** "
            "Camera coverage may be important for verifying "
            "the timeline.\n\n"
        )

        patterns_found = True

    if "phone" in lower_text or "wifi" in lower_text:

        output += (
            "📱 **Digital evidence detected:** "
            "Phone, Wi-Fi, or device information may help "
            "verify a person's location.\n\n"
        )

        patterns_found = True

    if "receipt" in lower_text or "card" in lower_text:

        output += (
            "💳 **Transaction evidence detected:** "
            "Payment records may help establish where a "
            "transaction occurred.\n\n"
        )

        patterns_found = True

    if "photo" in lower_text or "image" in lower_text:

        output += (
            "📸 **Visual evidence detected:** "
            "Photographs may contain additional information "
            "that should be compared with the written clues.\n\n"
        )

        patterns_found = True

    if not patterns_found:

        output += (
            "No specialized evidence type was automatically detected.\n\n"
        )

    output += "---\n\n"

    # --------------------------------------------------------
    # MOST LIKELY EXPLANATION
    # --------------------------------------------------------

    output += "# 🔎 MOST LIKELY EXPLANATION\n\n"

    if suspicious_lines:

        output += (
            "The strongest current hypothesis is that the case "
            "contains an unresolved inconsistency in the supplied "
            "evidence. The available clues are not sufficient to "
            "prove exactly what happened.\n\n"
        )

    else:

        output += (
            "The available evidence does not provide enough "
            "information to determine a single explanation.\n\n"
        )

    # --------------------------------------------------------
    # ALTERNATIVE EXPLANATIONS
    # --------------------------------------------------------

    output += "# 🔀 ALTERNATIVE EXPLANATIONS\n\n"

    output += (
        "1. One of the recorded statements may be inaccurate.\n"
        "2. A timestamp or description may have been recorded incorrectly.\n"
        "3. There may be additional evidence that has not been provided.\n"
        "4. Two apparently conflicting clues may describe different events.\n\n"
    )

    # --------------------------------------------------------
    # MISSING EVIDENCE
    # --------------------------------------------------------

    output += "# ❓ MISSING EVIDENCE\n\n"

    output += (
        "- Original photographs or videos\n"
        "- Complete CCTV footage\n"
        "- Exact timestamps\n"
        "- Device/location records\n"
        "- Witness statements\n"
        "- Original documents or transaction records\n\n"
    )

    # --------------------------------------------------------
    # CONFIDENCE
    # --------------------------------------------------------

    if len(lines) >= 8 and suspicious_lines:

        confidence = "MEDIUM"

    elif len(lines) >= 5:

        confidence = "LOW-MEDIUM"

    else:

        confidence = "LOW"

    output += "---\n\n"

    output += f"# 🎯 CONFIDENCE: {confidence}\n\n"

    output += (
        "This result is an analytical hypothesis, not proof. "
        "The system should not be used to accuse someone of "
        "wrongdoing without independent evidence.\n"
    )

    return output


# ============================================================
# DEMO MYSTERY
# ============================================================

DEMO_CASE = """
8:00 AM - John says he is at home.

8:15 AM - A message is sent from John's phone saying "I'm still at home."

8:25 AM - A coffee shop receipt shows a purchase using John's card at Central Station.

8:30 AM - CCTV shows someone who appears to be John entering Central Station.

8:35 AM - John's phone connects to Wi-Fi at his home.

9:00 AM - John says he left home at 8:20 AM.

The distance between John's home and Central Station is approximately 25 minutes by car.

John says he did not buy coffee.

The coffee shop receipt has John's card number but no photo.
"""


def run_demo():

    return solve_mystery(
        "The Impossible Journey",
        DEMO_CASE
    )


# ============================================================
# GRADIO WEBSITE
# ============================================================

with gr.Blocks(
    title="AI Mystery Solver"
) as app:

    gr.Markdown(
        """
# 🕵️ AI Mystery Solver

### Turn messy clues into timelines, contradictions and hypotheses.

Enter a mystery below and the system will analyze the evidence.
"""
    )

    with gr.Row():

        with gr.Column():

            case_title = gr.Textbox(
                label="🗂️ Case Title",
                value="My Mystery"
            )

            clues = gr.Textbox(
                label="📝 Enter Mystery Clues",
                placeholder="""
Example:

8:00 AM - John says he is at home.
8:30 AM - CCTV shows someone who appears to be John at the station.
8:35 AM - John's phone connects to Wi-Fi at home.
9:00 AM - John says he left home at 8:20 AM.
""",
                lines=18
            )

            with gr.Row():

                solve = gr.Button(
                    "🧠 SOLVE MYSTERY",
                    variant="primary"
                )

                demo_button = gr.Button(
                    "🎬 TRY DEMO"
                )

        with gr.Column():

            result = gr.Markdown(
                """
## 👋 Welcome

Enter a mystery on the left and click:

### 🧠 SOLVE MYSTERY

Or click:

### 🎬 TRY DEMO

**No API key is required.**
"""
            )

    solve.click(
        solve_mystery,
        inputs=[
            case_title,
            clues
        ],
        outputs=result
    )

    demo_button.click(
        run_demo,
        outputs=result
    )

    gr.Markdown(
        """
---

### ⚠️ Prototype Notice

This project identifies patterns and inconsistencies.
It does not establish guilt, criminal responsibility, or forensic truth.
"""
    )


# ============================================================
# LAUNCH
# ============================================================

print("🚀 Starting AI Mystery Solver...")
print("🌐 A public Gradio link will appear below.")

app.launch(
    share=True,
    debug=False
)
```
